```python
import numpy as np
```

```python
def numerical_grad(X,w,y,activation_fn,loss_fn):
    grad=np.empty_like(w)
    eps=1e-5
    for i in range(len(w)):
        e=np.zeros_like(w)
        e[i]=1*eps
        grad[i]=(loss_fn(y,activation_fn(X,w+e))-loss_fn(y,activation_fn(X,w-e)))/2*eps
    return grad

```

```python
def loss_weight(X, w, y, activation_fn, loss_fn, alpha):
    y_pred=activation_fn(X @ w)
    loss=loss_fn(y, y_pred)
    gradient=numerical_grad(X,w,loss_fn)
    w_new=w-alpha*gradient
    return loss, w_new
```

```python
def gradient_descent(X ,y,activation_fn, loss_fn, alpha,min_grad=1e-3, iter=1000):
    w=np.empty(X.shape[1])
    losses=[]
    for i in iter:
        loss, w_new=loss_weight(X, w, y, activation_fn, loss_fn, alpha, iter=1000)
        losses.append[loss]
        grad=w_new-w
        if np.linalg.norm(grad)<min_grad:
            break
        w=w_new
```

```python
def sigmoid(z):
    return np.where(z>0, 
                    1/(1+np.exp(-z)+1e-10), 
                    np.exp(z)/(1+np.exp(z)+1e-10)
                )
```

```python
def relu(z):
    return np.where(z>=0, z, 0)
```

```python
def BCE(y_pred, y):
    y_pred=np.clip(y_pred,1e-10,1-1e-10)
    return np.sum(y*np.log(y_pred)+(1-y)*np.log(1-y_pred))
```

```python
def hinge_loss(y_pred, y):
    return np.sum(relu(1 - y*y_pred))
```
